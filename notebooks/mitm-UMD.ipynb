# <strong>Man-in-the-Middle (MITM) Attacks - Setup Guide</strong>

<b>Step 1:</b> On SPHERE, navigate to the sidebar, and select <b>Experiments > Manage</b>. Click <b>New Experiment</b> in the upper righthand corner. In the Project dropdown menu, select your username. Give it the name `mitm`, and give it any description you want.

<b>Step 2:</b> On the sidebar, click <b>Model Editor</b>, then click "Start From Scratch."

<b>Step 3:</b> Copy/Paste the following code below, then paste it into the Model Editor.
```
from mergexp import *
net = Network('mitm', addressing==ipv4, routing==static)
alice,bob,eve = [net.node(name) for name in [ 'alice', 'bob', 'eve' ]]

# Create a link connecting the two nodes.
link = net.connect([alice,bob,eve])

# Making the experiment runnable.
experiment(net)
```
<b>Step 4:</b> At the top of the code editor, there is a "Play" button. Click on the button to the RIGHT of this, which says "Push to Experiment Repository" when you hover over it.

<b>Step 5:</b> Click on the Experiment dropdown, then click `mitm.umdclassXXXX`. Then, click "Push."

<b>Step 6:</b> Navigate to the sidebar, and select <b>Experiments > Manage</b>. Click on the `mitm` experiment that you selected earlier.

<b>Step 7:</b> Within the Models box, click on the name of your revision (it looks like a hash value). When you click on this, you should see the topology of three nodes, which show `alice`, `bob`, and `eve`.

<b>Step 8:</b> In the upper righthand corner, click "Reserve Revision." Give it the name `real`, and do not change anything else.

<b>Step 9:</b> Navigate to the sidebar, and select <b>Experiments > Reservations</b>. Scroll to the right of this table, and find the kebab menu (three vertical dots). Click on it, then select "Activate".

<b>Step 10:</b> Navigate to the sidebar, and select <b>Experiments > Activations</b>. Scroll to the right of this table, and find the <b>Status</b>. Wait until this is green, and says <b><span style="color: green">Success</span></b>. You can refresh after a minute has passed if it hasn't updated yet.

<b>Step 11:</b> Navigate to the sidebar, and select <b>XDCs</b>. Scroll to the right of this table, and find the kebab menu for this row. Click "Attach," then select the Materialization and select the one that says `real.mitm.umdclassXXXX`.

<b>Step 12:</b> Now, re-open your XDC, or refresh your browser tab (since you are already on it, reading this document).

<b>Step 13:</b> Open a new terminal, and run the following command: `bash /home/runlab mitm`

<b>Step 14:</b> Wait until the script ends. 

<b><span style="color: red">Important:</span></b> This script may look frozen. If you press "Return" and just see a new line, the script is still running. The script is complete when you see the following:
```
. . . . . . . . . . . . . . . . 
/etc/nologin not set. not deleting.
>>> All node installs finished.

umdclassXXXX@xdc-umdclassXXXX:~$
```

<b>Step 14:</b> Type `ssh eve`, then type `sudo ./install-ettercap.sh`. This will configure one more dependency for your lab. If you do not see `install-ettercap.sh` in the home directory, the `runlab` script failed.

<b>Step 15:</b> The setup is complete! To confirm that everything has installed correctly, try typing `ettercap`, `tcpdump`, and `chaosreader`. If any of these say `command not found`, there was an issue somewhere on Step 13 or Step 14. <u>Your lab is available for one week, so feel free to close this notebook and view the lab instruction manual.</u>
