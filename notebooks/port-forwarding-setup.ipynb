# <strong>Setting Up Port Forwarding</strong>

Three labs (Pathname, SQL Injection and Cross-Site Scripting) require port forwarding to be set up in order to access a web server that's hosted from these nodes. This notebook will be a guide that will assist you with how to configure port forwarding for your machine.

### Step 1: Download the `mrg` command line interface (CLI)

Navigate to <a href="https://gitlab.com/mergetb/portal/cli/-/releases/">this website</a> and download the appropriate file for your operating system. There are many releases of the CLI, which depends on your operating system.

- <strong>Windows</strong>: `mrg_X.X.XX_windows_amd64.tar.gz`
- <strong>MacOS (Intel)</strong>: `mrg_X.X.XX_darwin_amd64.tar.gz`
- <strong>MacOS (Apple Silicon)</strong>: `mrg_X.X.XX_darwin_arm64.tar.gz`
    - If you're unsure of your processor, click on the Apple menu, then "About This Mac". If your chip name starts with Apple, use the Apple Silicon version. Intel chips are only used in MacBooks that are from 2020 or older.
- <strong>Linux</strong>: `mrg_X.X.XX_linux_arm64.deb`

### Step 4: Configuring Your CLI

In the previous step, you were given a command to test your access to the CLI. Now, you will need to set the "configuration server" before port forwarding will work. Use the following command to set this up. <u>You only need to do this after the initial installation, and you won't need to do this again.</u>

- <strong>Windows:</strong> `./mrg.exe config set server grpc.sphere-testbed.net`
- <strong>MacOS:</strong> `./mrg config set server grpc.sphere-testbed.net`
- <strong>Linux:</strong> `mrg config set server grpc.sphere-testbed.net`

Next, you will need to sign into your SPHERE account. You can do this by running the following command: `mrg login USERNAME_GOES_HERE -p <password>`

<span style="color: red;"><strong><img src="resources/alert.png" style="width: 12px"> Important:</strong></span> The remaining examples of these commands will use the Linux syntax. Make sure that you keep track of using `./mrg.exe` if you're on Windows, and `./mrg` if you're on MacOS!

### Step 5: Performing a Port Forwarding Command

You should now be signed into the CLI and have access to port forwarding commands. These commands will be different, depending on the lab that you are working on. These commands will be provided in the notebook(s) when required, but for convenience, you will be using these commands for these three notebooks:

- <strong>Pathname</strong>: `mrg xdc ssh -x xdc.USERNAME_GOES_HERE -L 5000:5000 pathname`
- <strong>SQLi</strong>: `mrg xdc ssh -x xdc.USERNAME_GOES_HERE -L 8080:80 sqli`
- <strong>XSS</strong>: `mrg xdc ssh -x xdc.USERNAME_GOES_HERE -L 8080:80 server`

If you ever receive this error, it means that you must sign in again:

```
[user@user-ms7e06 ~]$ mrg whoami
Error:  Unknown
Message:        unable to retrieve identity: rpc error: code = NotFound desc = no access token provided
```

### Step 6: Working in Your Session

After running the command above, you now have direct access to your experimental node for the notebook. You may minimize your terminal and proceed working within JupyterLab. However, if you wish to work on your lab within your device's terminal, you're free to do so.

The terminal is only to bridge the connection from your experimental node to your device. Once you are done working on your lab, you may type `exit` in your terminal to suspend the port forwarding session. If you're working on a laptop, and close your screen, this will close the connection, and you will have to re-type the port forwarding command again.

Please refer to the notebook with how to access the website that's hosted from your experimental node.
